```python
df.groupby('Category')[['Quantity','Sales']].sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

#Grouping data
category_summary=df.groupby('Category')[['Quantity','Sales']].sum()


#plotting
fig, ax=plt.subplots(1,2,figsize=(14,5))


#bar plot for quantity
sns.barplot(x=category_summary.index,y=category_summary['Quantity'],ax=ax[0],palette='Blues')
ax[0].set_title('Total Quantity Sold by Product Category')
ax[0].set_xlabel('Product Category')
ax[0].set_ylabel('Total Quantity')
ax[0].tick_params(axis='x',rotation=45)

#bar plot for sales (assuming 'Sales' is the second column of interest)
sns.barplot(x=category_summary.index,y=category_summary['Sales'],ax=ax[1],palette='Greens')
ax[1].set_title('Total Sales by Product Category')
ax[1].set_xlabel('Product Category')
ax[1].set_ylabel('Total Sales')
ax[1].tick_params(axis='x',rotation=45)

plt.tight_layout()
plt.show()
```

Create a PieChart Showing the Percentage for each Category

```python
#count occurances of each product category
category_counts=df['Category'].value_counts()

#define explode to highlight the largest category
explode_list=[0.1 if i==category_counts.idxmax()else 0 for i in category_counts.index]

#plot pie chart
plt.figure(figsize=(10,7)) # Corrected figure height
plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct='%1.1f%%',  #display percentage values
    colors=plt.cm.Paired.colors,  #different for categories
    explode=explode_list,  #highlighting the highest value
    startangle=140,
    shadow=True

)



plt.title('Product Category Distribution')
plt.axis('equal') #ensures the pie chart is circular
plt.show()
```

Split the Dates into day,month and year

```python
#seeing the data column
df['Date']=pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
#add 'month' column and extract month from date
df['Month']=df['Date'].dt.month_name()
df['Month']
```

Total Sales Amount over each Month


```python
#sales trends overtime
plt.figure(figsize=(12,6))
sns.lineplot(x='Month',y='Sales',data=df) # Corrected 'Total Amount' to 'Sales'
plt.title('Sales trend Overtime',fontsize=16)
plt.xlabel('Month',fontsize=14)
plt.ylabel('Total Sales Amount per Month',fontsize=14)
plt.show()
```

```python
transaction_count=df.groupby('Month')['Order ID'].count()


plt.figure(figsize=(12,6))
sns.lineplot(x=transaction_count.index,y=transaction_count.values)
plt.title('Transaction Frequency Over Months',fontsize=16)
plt.xlabel('Month',fontsize=14)
plt.ylabel('Transaction_count',fontsize=14)
plt.show()
```

Transaction Frequency over each month


```python
plt.figure9figsize=(12,6)
sns.barplot(x=transaction_count.index,y=transaction_count.values)
sns.barplot(x=transaction_count.index,y=transaction_count.values)
plt.title('Transaction Frequency Over Months',fontsize=16)
plt.xlabel('Month',fontsize=14)
plt.ylabel('Transaction_count',fontsize=14)
plt.show()
```

Sales Heatmap


```python
heatmap=df.pivot_table(index='Month',columns='Category',values='Sales',aggfunc='sum')
plt.figure(figsize=(12,6))
sns.heatmap(heatmap,annot=True,cmap='YlGnBu')
plt.title('Sales Heatmap',fontsize=16)
plt.xlabel('Category',fontsize=14)
plt.ylabel('Month',fontsize=14)
plt.show()
```
